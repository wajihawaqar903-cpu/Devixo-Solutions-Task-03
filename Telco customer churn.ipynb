```python

# 1. Count missing values before handling them
print("Missing values before handling:")
print(df.isnull().sum())

# 2. Fill missing values
for column in df.columns:
    if df[column].isnull().sum() > 0:
        if pd.api.types.is_numeric_dtype(df[column]):
            # Fill numerical columns with median
            df[column] = df[column].fillna(df[column].median())
        else:
            # Fill categorical columns with most frequent value
            mode_value = df[column].mode()
            if not mode_value.empty:
                df[column] = df[column].fillna(mode_value[0])

# 3. Verify missing values after handling
print("\nMissing values after handling:")
print(df.isnull().sum())

print("\nTotal missing values remaining:", df.isnull().sum().sum())
```

```python

# 1. Count duplicate rows
duplicates = df.duplicated().sum()
print("Number of duplicate rows before removal:", duplicates)

# 2. Remove duplicate rows
df = df.drop_duplicates()

# 3. Verify duplicates are removed
print("Number of duplicate rows after removal:", df.duplicated().sum())

# 4. Display updated dataset shape
print("Dataset shape after removing duplicates:", df.shape)
```

```python

import pandas as pd

# 1. Drop useless columns
if "customerID" in df.columns:
    df = df.drop(columns=["customerID"])

print("Customer ID column removed (if it existed).")

# 2. Convert TotalCharges to numeric
# Some versions of this dataset contain blank strings in this column.
if "TotalCharges" in df.columns:
    df["TotalCharges"] = pd.to_numeric(
        df["TotalCharges"], errors="coerce"
    )
    df["TotalCharges"] = df["TotalCharges"].fillna(
        df["TotalCharges"].median()
    )

# 3. Create a tenure group
if "tenure" in df.columns:
    df["TenureGroup"] = pd.cut(
        df["tenure"],
        bins=[-1, 12, 24, 48, float("inf")],
        labels=["0-12 months", "13-24 months",
                "25-48 months", "49+ months"]
    )

# 4. Create charges per month feature
if "MonthlyCharges" in df.columns and "tenure" in df.columns:
    df["ChargesPerTenureMonth"] = (
        df["MonthlyCharges"] * df["tenure"]
    )

# 5. Convert categorical text columns into numeric columns
df = pd.get_dummies(df, drop_first=True, dtype=int)

# 6. Check the results
print("\nFirst 5 rows after feature engineering:")
display(df.head())

print("\nUpdated dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes.value_counts())

print("\nRemaining missing values:", df.isnull().sum().sum())
```

```python

from sklearn.feature_selection import SelectKBest, f_classif
import pandas as pd

# 1. Identify the target column
# Works with the standard Telco dataset after get_dummies()
if "Churn_Yes" in df.columns:
    target = "Churn_Yes"
elif "Churn" in df.columns:
    target = "Churn"
else:
    raise ValueError("Churn target column not found!")

# 2. Separate features (X) and target (y)
X = df.drop(columns=[target])
y = df[target]

# If the target is still text, convert it to numbers
if y.dtype == "object":
    y = y.map({"No": 0, "Yes": 1})

# Ensure all input features are numeric
X = X.select_dtypes(include="number").fillna(0)

# 3. Select the top 15 features (or fewer if unavailable)
k = min(15, X.shape[1])
selector = SelectKBest(score_func=f_classif, k=k)
X_selected_array = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

# 4. Display feature scores
scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
}).sort_values(by="Score", ascending=False)

print("Feature scores (highest first):")
display(scores)

# 5. Keep only the selected features
X_selected = pd.DataFrame(
    X_selected_array,
    columns=selected_features,
    index=X.index
)

print(f"\nSelected {k} features:")
print(selected_features.tolist())

print("\nSelected features shape:", X_selected.shape)
print("Target shape:", y.shape)
```

```python

from sklearn.model_selection import train_test_split

# Split selected features into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the results
print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nTraining data percentage:",
      round(len(X_train) / len(X_selected) * 100, 2), "%")
print("Testing data percentage:",
      round(len(X_test) / len(X_selected) * 100, 2), "%")
```

```python

from sklearn.preprocessing import StandardScaler

# 1. Create the scaler
scaler = StandardScaler()

# 2. Fit on training data only and transform it
X_train_scaled = scaler.fit_transform(X_train)

# 3. Transform testing data using the same scaler
X_test_scaled = scaler.transform(X_test)

# 4. Display the results
print("Original training shape:", X_train.shape)
print("Scaled training shape:", X_train_scaled.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Scaled testing shape:", X_test_scaled.shape)

print("\nData scaling completed successfully!")
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

# Define the four models
models = {
    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ),

    "Support Vector Machine (SVC)": SVC(
        class_weight="balanced",
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}

# Train each model and display its results
trained_models = {}

for name, model in models.items():
    print(f"\n{'=' * 50}")
    print(f"Training: {name}")

    # SVC and the other models can use scaled data
    model.fit(X_train_scaled, y_train)

    # Save the trained model
    trained_models[name] = model

    # Predict on test data
    y_pred = model.predict(X_test_scaled)

    # Display evaluation results
    print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

print("\nAll four models have been trained successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

# Store evaluation results
evaluation_results = []

for name, model in trained_models.items():

    # Make predictions
    y_pred = model.predict(X_test_scaled)

    # Get probabilities or decision scores for ROC-AUC
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_test_scaled)[:, 1]
    else:
        y_score = model.decision_function(X_test_scaled)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )
    roc_auc = roc_auc_score(y_test, y_score)

    evaluation_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc
    })

    # Print classification report
    print("\n" + "=" * 60)
    print("MODEL:", name)
    print("=" * 60)
    print(classification_report(
        y_test, y_pred, zero_division=0
    ))
    print("ROC-AUC:", round(roc_auc, 4))

    # Plot confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No Churn", "Churn"]
    )
    disp.plot(cmap="Blues", values_format="d")
    plt.title(f"Confusion Matrix - {name}")
    plt.tight_layout()
    plt.show()

# Create a comparison table
evaluation_df = pd.DataFrame(evaluation_results)
evaluation_df = evaluation_df.sort_values(
    by="F1-Score", ascending=False
).reset_index(drop=True)

print("\nFINAL MODEL EVALUATION COMPARISON")
display(evaluation_df.round(4))
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Display comparison table of all model scores
print("MODEL PERFORMANCE COMPARISON")
display(
    evaluation_df
    .sort_values(by="F1-Score", ascending=False)
    .reset_index(drop=True)
    .round(4)
)

# 2. Prepare data for visualization
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

plot_df = evaluation_df.set_index("Model")[metrics]

# 3. Create grouped bar chart
ax = plot_df.plot(
    kind="bar",
    figsize=(13, 6),
    width=0.8
)

plt.title("Performance Comparison of Churn Prediction Models")
plt.xlabel("Machine Learning Model")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=20, ha="right")
plt.legend(title="Evaluation Metrics", bbox_to_anchor=(1.02, 1))
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report
)

# 1. Define the model
rf = RandomForestClassifier(
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

# 2. Define parameters to test
param_dist = {
    "n_estimators": [50, 100],
    "max_depth": [10, 20, None],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

# 3. Search fewer combinations using 3-fold cross-validation
search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=6,
    cv=3,
    scoring="f1",
    random_state=42,
    n_jobs=1,
    verbose=2
)

# 4. Tune using training data only
print("Starting hyperparameter tuning...")
search.fit(X_train_scaled, y_train)

# 5. Display best parameters
best_rf = search.best_estimator_

print("\nBest Parameters:")
print(search.best_params_)

print("\nBest Cross-Validation F1-Score:")
print(round(search.best_score_, 4))

# 6. Evaluate on the test data
y_pred = best_rf.predict(X_test_scaled)
y_prob = best_rf.predict_proba(X_test_scaled)[:, 1]

print("\nTuned Random Forest Results")
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(
    precision_score(y_test, y_pred, zero_division=0), 4
))
print("Recall:", round(
    recall_score(y_test, y_pred, zero_division=0), 4
))
print("F1-Score:", round(
    f1_score(y_test, y_pred, zero_division=0), 4
))
print("ROC-AUC:", round(roc_auc_score(y_test, y_prob), 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python

from sklearn.model_selection import cross_val_score

# Perform 5-fold cross-validation on the best model
cv_scores = cross_val_score(
    best_rf,
    X_train_scaled,
    y_train,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

# Display individual fold scores
print("5-Fold Cross-Validation F1 Scores:")
print(cv_scores)

# Calculate and display average score
print("\nAverage Cross-Validation F1-Score:",
      round(cv_scores.mean(), 4))

# Display standard deviation
print("Standard Deviation:",
      round(cv_scores.std(), 4))
```
